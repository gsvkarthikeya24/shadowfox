```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set_theme(style="whitegrid")
import numpy as np
```

## **Pandas Overview**:
Pandas is an open-source Python library for data manipulation and analysis.

It provides powerful, flexible tools to handle structured (table-like) data easily.
#**Unique Features of Pandas**
1.Data Structures:
*Series: 1-dimensional labeled array (like a column).

DataFrame: 2-dimensional labeled data structure (like a table).

Handling Missing Data:

Functions like .isnull(), .dropna(), and .fillna() make it easy to detect, remove, or replace missing values.

Data Alignment and Indexing:

Powerful tools for automatic data alignment and intelligent indexing

# **Matplotlib Overview:**
Matplotlib is a popular Python plotting library used to create static, animated, and interactive visualizations.
* It is highly customizable and works well with NumPy and Pandas data

# **Unique Features of Matplotlib:**
Wide Range of Plots:
Supports line plots, bar charts, scatter plots, histograms, pie charts, 3D plots, and more.

Highly Customizable:
Almost every part of a figure (colors, labels, ticks, legends) can be modified.

Publication-Quality Figures:
Create high-quality plots that are suitable for research papers and presentations.

# **NumPy Overview:**
NumPy (Numerical Python) is a fundamental Python library for scientific computing.
*It provides fast, efficient tools for working with arrays and matrices.

# **Unique Features of NumPy:**
N-Dimensional Array (ndarray):

*Core feature is the powerful n-dimensional array object for data storage and computation.

*Vectorized Operations:
Perform operations on entire arrays without writing loops, leading to faster and cleaner code.

*Broadcasting:
Automatically adjusts arrays of different shapes during operations, making code more flexible.

# **Seaborn Overview:**

*Seaborn is a Python data visualization library based on Matplotlib.
*It makes it easy to create beautiful, attractive, and informative statistical graphics.

# **Unique Features of Seaborn:**

*Built-in Themes and Styles:
Automatically applies attractive themes for better aesthetics compared to Matplotlib.

*Simplified Syntax:
Very easy to create complex plots with minimal code.

*High-Level Interface:
Functions like sns.relplot(), sns.catplot(), sns.pairplot() simplify complex visualizations.

```python
df=pd.read_csv("/content/commodity_prices (1) (1).csv")
df.head()
```

```python
print(df['coffee_robustas'])
```

```python

plt.figure(figsize=(15,5))
sugar_plot = sns.lineplot(data=df[["sugar_eu"]].rolling(20).mean())
plt.title("sugar")
plt.ylabel("prce per kilogram")
plt.xlabel("date")
h,_=sugar_plot.get_legend_handles_labels()
sugar_plot.legend(h,["european union"], loc="upper left")
```

```python
plt.figure(figsize=(15,6))
oil_plot =sns.lineplot(data=df[["oil_brent","oil_dubai"]].rolling(20).mean())
plt.title("oil_prices")
plt.ylabel("pice per barrel")
plt.xlabel("date")
h,__=oil_plot.get_legend_handles_labels()
```
